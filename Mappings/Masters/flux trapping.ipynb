```python
import numpy as np
import scipy.constants as sc
```

```python

def lower_field(Phi_0, W, dzeta, labda):
    return Phi_0/(4*np.pi*labda**2)*(np.log(labda/dzeta)+.5)

def lower_field_narrow_strip(Phi_0, W, dzeta):
    return 2*Phi_0/(np.pi*W**2)*np.log(2*W/(np.pi*dzeta))

def labda_eff(labda_L, dzeta_P, d):
    return labda_L*np.sqrt(dzeta_P/d)

def labda_perp(labda_eff, d):
    return labda_eff**2/d



d = 40e-9   # nm
w = 4e-6   # nm
labda = 6.5e-5   # nm
dzeta = 19e-9 # nm
Phi_0 = np.pi * sc.h/sc.e

mu0_Hc1 = lower_field(Phi_0, w, dzeta, labda)
mu0_Hc1_w = lower_field_narrow_strip(Phi_0, w, dzeta)
print("mu0_Hc1 = ", mu0_Hc1*1e6, "muT")
print("mu0_Hc1 strip = ", mu0_Hc1_w*1e6, "muT")
```
